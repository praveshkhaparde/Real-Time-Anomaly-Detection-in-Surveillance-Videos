```python
import os
import random
from IPython.display import display, Image
from PIL import Image as PILImage
import matplotlib.pyplot as plt

# Sample a few image names from original set
image_folder = "/content/yolov5/datasets/coco128/images/train2017"
pred_folder = "/content/yolov5/runs/detect/exp2"

image_files = os.listdir(image_folder)
sample_images = random.sample(image_files, 3)

# Show original and predicted side-by-side
for img_name in sample_images:
    orig_path = os.path.join(image_folder, img_name)
    pred_path = os.path.join(pred_folder, img_name)

    fig, ax = plt.subplots(1, 2, figsize=(12, 6))
    ax[0].imshow(PILImage.open(orig_path))
    ax[0].set_title("Original Image")
    ax[0].axis("off")

    ax[1].imshow(PILImage.open(pred_path))
    ax[1].set_title("YOLOv5 Prediction")
    ax[1].axis("off")

    plt.show()
```
